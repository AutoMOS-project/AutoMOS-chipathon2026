```python
import pygmid
print(pygmid.__version__)
from pygmid import Lookup as lk
import pandas as pd
import numpy as np
```

```python
import os
os.listdir('/foss/designs/Chipathon2025/lookup_tables')
```

```python
n = lk('/foss/designs/Chipathon2025/lookup_tables/nfet_03v3.mat')
p = lk('/foss/designs/Chipathon2025/lookup_tables/pfet_03v3.mat')
```

```python
Ls = [0.28, 0.5, 1, 2, 4]
id =  1e-6
gm_id = np.array([5, 10, 15, 20, 25])
gm = gm_id*id
```

```python
dfs = {}

for l in Ls:
    jd = n.lookup('ID_W', GM_ID=gm_id, L=l)
    w = id/jd
    cgg_w = n.lookup('CGG_W', GM_ID=gm_id, L=l)
    cgg = w*cgg_w
    ft = gm/cgg/(2*np.pi)
    df = pd.DataFrame([gm_id, gm, jd, w, cgg, ft], ['gm_id', 'gm','jd', 'w', 'cgg', 'ft'])
    dfs[l] = df
```
