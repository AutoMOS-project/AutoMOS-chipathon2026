```python
id_deseada = 4e-6
L = 4
VDS = 1.65
GM_ID_target = 12

# Rango más amplio de VGS
vgs_range = np.linspace(0, 3.3, 200)

# Listas para almacenar resultados
resultados = []
vgs_encontrado = None
w_encontrado = None
gm_id_encontrado = None
diferencia_minima = float('inf')

for vgs in vgs_range:
    try:
        # Buscar ID/W para este VGS
        jd = p.lookup('ID_W', VGS=vgs, L=L, VDS=VDS)
        w = id_deseada / jd
        
        # Calcular GM_ID para este punto
        gm = p.lookup('GM', VGS=vgs, L=L, VDS=VDS)
        gm_id = gm / jd
        
        # Guardar para depuración
        resultados.append((vgs, gm_id, w))
        
        # Calcular diferencia con el target
        diff = abs(gm_id - GM_ID_target)
        
        # Si es mejor que lo anterior, guardar
        if diff < diferencia_minima:
            diferencia_minima = diff
            vgs_encontrado = vgs
            w_encontrado = w
            gm_id_encontrado = gm_id
        
        # Si cumple con la tolerancia, podemos parar
        if diff < 0.5:  # tolerancia más amplia
            vgs_encontrado = vgs
            w_encontrado = w
            gm_id_encontrado = gm_id
            break
            
    except Exception as e:
        # print(f"Error en VGS={vgs:.3f}: {e}")  # Descomentar para depuración
        continue

# Mostrar resultados
if vgs_encontrado is not None:
    print(f"✅ VGS encontrado = {vgs_encontrado:.3f} V")
    print(f"✅ GM_ID obtenido = {gm_id_encontrado:.2f} (target: {GM_ID_target})")
    print(f"✅ W calculado = {w_encontrado:.3e} m")
    print(f"✅ Diferencia = {diferencia_minima:.3f}")
else:
    print("❌ No se encontró ningún VGS en el rango")
    print("📊 Revisa si los valores de lookup existen para L=2, VDS=0.65")

# Opcional: Mostrar los primeros resultados para depuración
print("\n📋 Primeros valores explorados:")
for i in range(min(5, len(resultados))):
    v, g, ww = resultados[i]
    print(f"  VGS={v:.3f}, GM_ID={g:.2f}, W={ww:.3e}")
```

```python
def find_vds(n, id_deseada, L, VGS, GM_ID_target, tol=0.5):
    """Encuentra VDS para un VGS y GM_ID dados"""
    for vds in np.linspace(0.05, 1.8, 200):
        try:
            jd = p.lookup('ID_W', VGS=VGS, L=L, VDS=vds)
            gm = p.lookup('GM', VGS=VGS, L=L, VDS=vds)
            if abs(gm/jd - GM_ID_target) < tol:
                return vds, id_deseada / jd
        except:
            continue
    return None, None

# Uso
vds, w = find_vds(p, 1e-6, 4, 2, 5)
if vds:
    print(f"VDS = {vds:.3f} V, W = {w:.3e} m")
else:
    print("No encontrado")
```
